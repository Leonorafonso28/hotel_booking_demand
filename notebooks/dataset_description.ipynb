# **Research Question**

Are there hidden behavioural clusters among travel agencies and tour operators when jointly analysing distribution channels (DistributionChannel), booking lead time (LeadTime), and the volume of booking modifications (BookingChanges)?

## **Decision relevance**

In the hotel booking ecosystem, travel agencies and tour operators often exhibit heterogeneous booking behaviours that are not directly observable through individual variables alone. Clustering methods are particularly suitable in this context because they allow the discovery of hidden structures in unlabeled data, revealing groups of reservations that share similar behavioural characteristics. Understanding these latent clusters can provide valuable insights into how different intermediaries interact with the booking system. 

From a managerial perspective, discovering these behavioural segments can support several strategic decisions. First, it can improve channel management strategies, enabling hotels to tailor partnerships, incentives, or commission structures according to the behavioural profile of intermediaries. Second, it can support operational planning, since clusters characterised by high booking volatility or frequent changes may require more flexible inventory management. Finally, it can contribute to revenue management practices, as different booking behaviours may be associated with distinct pricing sensitivities, planning horizons, or risk profiles.

# **Files description**

- **hotel_bookings_course_release_v1.csv:** main dataset containing 119,390 records and 32 variables, where each row represents a hotel booking. The data cover arrivals between July 2015 and August 2017 across two Portuguese hotels (a resort in the Algarve and a city hotel in Lisbon). It includes variables such as room type, number of guests, distribution channel, ADR (Average Daily Rate), and booking status;

- **referencePaper.pdf:** the original scientific article published in the journal Data in Brief, which describes and documents the dataset. It explains the origin of the data, the feature engineering process, and provides detailed descriptive statistics for both hotels;

- **column_roles.csv:** annotations describing the role of each variable, mainly for leakage control. It indicates which variables are outcome/post-event variables and therefore must not be used as clustering inputs, and which variables have high cardinality;

- **subsample_indices_v1_n30000_seed12345.txt:** file containing row indices (0-based) for a reproducible subsample of 30,000 records, generated with seed 12345. It is intended for use with computationally intensive methods, ensuring reproducibility;

- **COURSE_RELEASE_FILES_README.txt:** instruction file for the dataset package, explaining the purpose of each file and the governance policies;

- **DATASET_MANIFEST.yml:** machine-readable data card containing dataset metadata, including version, dimensions, source, license (CC BY 4.0), and SHA-256 hashes for integrity verification;

- **SHA256SUMS.txt:** file containing SHA-256 cryptographic hashes for the main files, used to verify that local copies are complete and identical to the official course release version.
